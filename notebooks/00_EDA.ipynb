```python
import pandas as pd
```

# Data Exploration

```python
train_processed_path = '../data/features/processed/train_data_processed.csv'
test_processed_path = '../data/features/processed/test_data_processed.csv'

train_data = pd.read_csv(train_processed_path)
test_data = pd.read_csv(test_processed_path)
```

## Class distribution

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

train_data['label'].value_counts().plot(kind='bar', ax=axes[0])
axes[0].set_title('Train Set Class Distribution')
axes[0].set_xlabel('Class')
axes[0].set_ylabel('Count')

test_data['label'].value_counts().plot(kind='bar', ax=axes[1])
axes[1].set_title('Test Set Class Distribution')
axes[1].set_xlabel('Class')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
